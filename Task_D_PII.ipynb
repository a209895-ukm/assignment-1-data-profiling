## 1. Missing-Value Profiling

We count missing values and calculate their percentage in each
column before deciding how to handle them. These checks cover
the full customer and product tables.

```python
def missing_summary(df):
    summary = pd.DataFrame({
        "Missing Count": df.isna().sum(),
        "Missing (%)": (df.isna().mean() * 100).round(2),
    })

    return summary.sort_values("Missing Count", ascending=False)


customer_missing = missing_summary(customers)
product_missing = missing_summary(products)

print("CUSTOMERS")
display(customer_missing)

print("PRODUCTS")
display(product_missing)
```

### Customer Missing-Value Findings

Job Title is missing for 584,185 customers (35.55%).
All other customer columns have no missing values recognised
by pandas at this stage.

Proposed handling: retain these customer records and leave
unrecorded occupations as missing. Job Title is not essential
for the planned comparisons by age, gender, and product category.
Dropping these customers would unnecessarily reduce coverage.

If occupation is analysed later, missing occupations should be
reported as a separate "Not recorded" group, rather than guessed.

### Investigating Missing Product Attributes

Color is missing for 12,445 products (69.37%), while Sizes is
missing for 2,070 products (11.54%).

We examine missingness by product subcategory before deciding
how to handle these values. Missing values alone do not establish
whether an attribute is unknown or not applicable.

```python
product_missing_by_group = (
    products
    .assign(
        color_missing=products["Color"].isna(),
        sizes_missing=products["Sizes"].isna(),
    )
    .groupby(["Category", "Sub Category"])
    .agg(
        product_count=("Product ID", "size"),
        color_missing_pct=("color_missing", "mean"),
        sizes_missing_pct=("sizes_missing", "mean"),
    )
)

percent_columns = ["color_missing_pct", "sizes_missing_pct"]

product_missing_by_group[percent_columns] = (
    product_missing_by_group[percent_columns] * 100
).round(2)

display(product_missing_by_group)
```

### Product Missing-Value Decisions

All 2,070 missing Sizes values occur in Accessories, with 690
affected products in each main category. No other subcategory
has missing Sizes. This suggests a category-related recording
pattern, but does not confirm that all accessories are one-size.

Color is missing for 69.37% of products, with missingness varying
substantially across subcategories. Dropping these records would
remove some subcategories entirely and distort category coverage.

We retain all products, preserve the missing values, and add
missingness flags. We do not impute colours or sizes because
there is insufficient evidence to determine their true values.
Comparisons involving these attributes must report this limitation.

```python
# Create working copies, keeping the original tables unchanged.
customers_clean = customers.copy()
products_clean = products.copy()

# Flag missing information without guessing replacement values.
customers_clean["Job Title Missing"] = (
    customers_clean["Job Title"].isna()
)

products_clean["Color Missing"] = (
    products_clean["Color"].isna()
)

products_clean["Sizes Missing"] = (
    products_clean["Sizes"].isna()
)

print("Missing-value flags:")
print("Job Title:", customers_clean["Job Title Missing"].sum())
print("Color:", products_clean["Color Missing"].sum())
print("Sizes:", products_clean["Sizes Missing"].sum())

print("\nRows retained:")
print("Customers:", len(customers_clean))
print("Products:", len(products_clean))
```

### Full Transaction Missing-Value Profiling

We process the transactions file in chunks of 100,000 rows and
accumulate missing-value counts across all chunks. This checks
the entire file rather than only the first 1,000 records.

```python
transaction_rows = 0
transaction_missing_counts = None

for chunk_number, chunk in enumerate(
    pd.read_csv(
        "data/transactions.csv",
        chunksize=100_000,
    ),
    start=1,
):
    chunk_missing = chunk.isna().sum()

    if transaction_missing_counts is None:
        transaction_missing_counts = chunk_missing
    else:
        transaction_missing_counts = (
            transaction_missing_counts + chunk_missing
        )

    transaction_rows += len(chunk)

    # Print occasional progress updates.
    if chunk_number % 10 == 0:
        print(f"Checked {transaction_rows:,} rows...")

transaction_missing = pd.DataFrame({
    "Missing Count": transaction_missing_counts,
    "Missing (%)": (
        transaction_missing_counts / transaction_rows * 100
    ).round(2),
}).sort_values("Missing Count", ascending=False)

print(f"\nTotal transaction rows checked: {transaction_rows:,}")
display(transaction_missing)
```

# Task C: Outliers and Diagnostic Charts

Task B already confirmed that no impossible values exist (no negative prices or costs, no quantity ≤ 0, discounts within 0–1, no birth dates after the last transaction). Here we look for values that are unusual but possibly genuine, using the IQR rule: values below Q1 − 1.5×IQR or above Q3 + 1.5×IQR are flagged as outliers.

# Age

Age is not stored in the data, so we calculate it from Date Of Birth at the date of the last transaction (18 March 2025), so that age reflects the time the data was recorded.

```python
ref_date = transactions["Date"].max()
customers_clean["Age"] = (ref_date - customers_clean["Date Of Birth"]).dt.days / 365.25
print("Reference date:", ref_date)
display(customers_clean["Age"].describe())
print("Customers under 18:", (customers_clean["Age"] < 18).sum())
```

Ages range from 18.0 to 76.0 years (mean 31.9). No customer is under 18 and none is above 100, so there are no impossible ages.

# Outlier Detection (IQR)

We apply the IQR rule to every numerical field on the full data. Line Total is split into sales and returns using Transaction Type, because return amounts are negative and would distort the sales quartiles.

```python
def iqr_bounds(s):
    q1, q3 = s.quantile([0.25, 0.75])
    iqr = q3 - q1
    return q1 - 1.5 * iqr, q3 + 1.5 * iqr

is_return = transactions["Transaction Type"] == "Return"
checks = {
    "Unit Price": transactions["Unit Price"],
    "Quantity": transactions["Quantity"],
    "Discount": transactions["Discount"],
    "Line Total (sales)": transactions.loc[~is_return, "Line Total"],
    "Line Total (returns)": transactions.loc[is_return, "Line Total"],
    "Age": customers_clean["Age"],
    "Production Cost": products_clean["Production Cost"],
}
rows = []
for name, s in checks.items():
    low, high = iqr_bounds(s)
    n = ((s < low) | (s > high)).sum()
    rows.append([name, len(s), round(low, 2), round(high, 2), n, round(n / len(s) * 100, 2)])

outlier_summary = pd.DataFrame(
    rows, columns=["column", "n_rows", "lower", "upper", "n_outliers", "pct_outliers"]
)
outlier_summary
```

Many values are flagged because retail prices are right-skewed. Unit Price has 18.67% flagged, but the values form a continuous tail up to 1,153.5, which indicates genuine premium products. Quantity is flagged at 6.68% only because most lines have quantity 1 (IQR = 0), so quantities 2 and 3 are normal purchases. Discount has no outliers. Line Total follows unit price for both sales (15.32%) and returns (15.33%). Age (0.62%) and Production Cost (4.67%) are plausible. So the decision is to keep all, because no value is impossible and removing them would bias the return-rate analysis.

# Treatment: Capping Unit Price

To limit the influence of the most expensive products on summary statistics, Unit Price is capped at the 99th percentile in a new column. The original column is kept, and an `Is Price Outlier` flag is added so no information is lost.

```python
p99 = transactions["Unit Price"].quantile(0.99)
_, price_high = iqr_bounds(transactions["Unit Price"])

transactions["Unit Price (Capped)"] = transactions["Unit Price"].clip(upper=p99)
transactions["Is Price Outlier"] = transactions["Unit Price"] > price_high

print("99th percentile cap:", round(p99, 2))
print("Rows capped:", (transactions["Unit Price"] > p99).sum())
print("Rows flagged (IQR):", transactions["Is Price Outlier"].sum())
print("Rows before:", len(transactions), "| after:", len(transactions), "(no rows removed)")

pd.concat({"before": transactions["Unit Price"].describe(),
           "after (capped)": transactions["Unit Price (Capped)"].describe()}, axis=1)
```

The cap is 851.5, affecting 64,026 rows (1.00%). The maximum falls from 1,153.5 to 851.5 and the mean from 132.46 to 131.22, while the median and quartiles are unchanged. No rows were removed (6,416,029 before and after). Use the original Unit Price when calculating the value of returns, use the capped column only for charts and summary statistics.

## Task 5 : Privacy (PII) and Bias Risks

### 5.1 Personally Identifiable Information (PII) Identification and Handling
* **Direct PII Removal:** Direct identifiers (`Name`, `Email`, `Telephone`) were permanently removed to prevent re-identification risks and eliminate unnecessary identity tracking.
* **Quasi-Identifier Generalization:** `Date Of Birth` was transformed into discrete `Age_Group` brackets (`<18`, `18-24`, `25-34`, `35-49`, `50-64`, `65+`). Binning reduces the risk of re-identification through cross-referencing with external demographic tables while preserving actionable cohort trends.
* **Attribute Imputation:** Missing values in `Job Title` were filled with `'Unrecorded'` to avoid sample bias from listwise deletion.

| Field Name | PII Classification | Risk Exposure | Proposed Handling Strategy |
|---|---|---|---|
| **Name** | Direct Identifier | High (Personal identity exposure) | Drop field from analytical datasets; retain only anonymized `Customer ID`. |
| **Email** | Direct Identifier | High (Contact info / Account mapping) | Drop field because it's not required for demographic return segment analysis. |
| **Telephone** | Direct Identifier | High (Contact info / Device tracing) | Drop field; unstandardized formats are excluded from statistical pipelines. |
| **Date Of Birth** | Quasi-Identifier | Medium (Exact birthdate matching) | Generalize into discrete age brackets (e.g., 18–24, 25–34, 35–49, 50+). |
| **City & Country**| Quasi-Identifier | Low-Medium (Regional profiling) | Aggregate at country/regional level; suppress low-density city combinations. |
| **Job Title** | Quasi-Identifier | Low-Medium (35.55% missing) | Retain missing values as a distinct `"Unrecorded"` category. |

```python
# 5.1 Execution: PII Removal & Quasi-Identifier Handling
import pandas as pd
import numpy as np

# Ensure input dataset is loaded if running this cell independently
if 'df' not in globals() and 'df_clean' not in globals():
    df = pd.read_csv('data/customers.csv', dtype={'Telephone': 'string'}, low_memory=False)
elif 'df' not in globals() and 'df_clean' in globals():
    df = df_clean

# 1. Drop Direct PII Fields
pii_columns = ['Name', 'Email', 'Telephone']
df_clean = df.drop(columns=[col for col in pii_columns if col in df.columns])

# 2. Generalize 'Date Of Birth' into Age Brackets
current_year = 2026
df_clean['Age'] = current_year - pd.to_datetime(df_clean['Date Of Birth'], errors='coerce').dt.year

bins = [0, 17, 24, 34, 49, 64, 100]
labels = ['<18', '18-24', '25-34', '35-49', '50-64', '65+']
df_clean['Age_Group'] = pd.cut(df_clean['Age'], bins=bins, labels=labels)

# 3. Handle Missing Values for 'Job Title'
df_clean['Job Title'] = df_clean['Job Title'].fillna('Unrecorded')

print("PII Cleaning Complete. Updated Column List:")
print(df_clean.columns.tolist())
```

```python
# 5.3 Execution: Currency Standardization & Normalized Return Rate Calculation 
import gc
import numpy as np
import pandas as pd

# 1. Clear any leftover heavy variables from previous failed runs
for var_name in ['df_clean', 'analysis_df', 'merged_df']:
    if var_name in globals():
        del globals()[var_name]
gc.collect()

# 2. Load ONLY necessary transaction columns
trans_cols = [
    'Customer ID',
    'Product ID',
    'Transaction Type',
    'Unit Price',
    'Quantity',
    'Currency',
]
transactions = pd.read_csv('data/transactions.csv', usecols=trans_cols)

# Currency Standardization to USD Baseline
exchange_rates = {'USD': 1.0, 'EUR': 1.08, 'GBP': 1.27, 'CNY': 0.14}
transactions['Unit_Price_USD'] = transactions['Unit Price'] * transactions[
    'Currency'
].map(exchange_rates)
transactions['Line_Total_USD'] = (
    transactions['Quantity'] * transactions['Unit_Price_USD']
)

# 3. Aggregation-First: Summarize 6.4M transactions into small customer counts BEFORE merging
print('Aggregating transactions by customer...')
cust_summary = (
    transactions.groupby(['Customer ID', 'Transaction Type'], observed=False)
    .size()
    .unstack(fill_value=0)
)

# Load ONLY Customer ID and Gender from customers.csv
customers = pd.read_csv(
    'data/customers.csv', usecols=['Customer ID', 'Gender']
)

# Merge small aggregated summary table with customer demographic table
gender_merged = customers.merge(cust_summary, on='Customer ID', how='inner')

# Calculate Normalized Return Rate by Gender
gender_return = gender_merged.groupby('Gender')[['Sale', 'Return']].sum()
total_lines = gender_return['Sale'] + gender_return['Return']
gender_return['Return_Rate_%'] = np.where(
    total_lines > 0, (gender_return['Return'] / total_lines) * 100, 0
)

print('\nNormalized Return Rate by Gender (%):')
print(gender_return[['Sale', 'Return', 'Return_Rate_%']])

# Clean up intermediate memory
del cust_summary, gender_merged, customers
gc.collect()

# 4. Segment-Specific Sizing Missingness Audit (Directly on Products table)
products = pd.read_csv('data/products.csv')
subcat_col = (
    'Sub Category' if 'Sub Category' in products.columns else 'Subcategory'
)
size_col = 'Sizes' if 'Sizes' in products.columns else 'Size'

if subcat_col in products.columns and size_col in products.columns:
    apparel_products = products[products[subcat_col] != 'Accessories']
    missing_size_apparel = apparel_products[size_col].isnull().mean() * 100
    print(
        f'\nPercentage of Missing Size Data in Apparel Segment: {missing_size_apparel:.2f}%'
    )

# Clean up final variables
del transactions, products
gc.collect()
```

## 6. Conclusion and Fitness for Purpose

```python
# 6.1 Execution: Data Integrity & Quality Audit (Chunked Low-Memory)
import gc
import pandas as pd

# 1. Missingness Audit across Source Tables (customers & products)
customers = pd.read_csv(
    'data/customers.csv', usecols=['Gender', 'Country', 'Date Of Birth']
)
products = pd.read_csv('data/products.csv')

size_col = 'Sizes' if 'Sizes' in products.columns else 'Size'
color_col = 'Color' if 'Color' in products.columns else 'Colors'

print('Missingness Summary - Demographic Data (customers.csv):')
print(customers[['Gender', 'Country', 'Date Of Birth']].isnull().mean() * 100)

print('\nMissingness Summary - Product Attributes (products.csv):')
print(products[[color_col, size_col]].isnull().mean() * 100)

# Release memory from source tables
del customers, products
gc.collect()

# 2. Chunked Audit for Invoice Line Conflicts (Prevents OOM Spikes)
print('\nAuditing Invoice Line Conflicts ...')

chunk_list = []
# Stream transactions in 1-million-row chunks to keep RAM under 150MB
for chunk in pd.read_csv(
    'data/transactions.csv',
    usecols=['Invoice ID', 'Line', 'Product ID'],
    chunksize=1000000,
):
    # Deduplicate within each chunk immediately
    dedup_chunk = chunk.drop_duplicates(
        subset=['Invoice ID', 'Line', 'Product ID']
    )
    chunk_list.append(dedup_chunk)

# Combine lightweight deduplicated chunks
dedup_all = pd.concat(chunk_list, ignore_index=True).drop_duplicates(
    subset=['Invoice ID', 'Line', 'Product ID']
)

# Count conflicts: (Invoice ID, Line) combination tied to more than one Product ID
flagged_conflicts = dedup_all.duplicated(subset=['Invoice ID', 'Line']).sum()

print(f'Total Flagged Invoice Line Conflicts Detected: {flagged_conflicts}')

# Final memory cleanup
del chunk_list, dedup_all
_ = gc.collect()
```
